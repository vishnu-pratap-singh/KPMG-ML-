# missing values handling

```python
# replace missing values of age with median

df['age'] = df['age'].fillna(df['age'].median())
```

```python
# replace missing values of income with median
df['income']= df['income'].fillna(df['income'].median())
```

```python
# # replace missing values of employment with mode

df['employment']= df['employment'].fillna(df['employment'].mode()[0])
```

# label Encoding 

```python
# label encoding 

categorical_cols = [
    'employment',
    'previous default',
    'credit history',
    'creditworthiness'
]

for col in categorical_cols:
    le=LabelEncoder()
    df[col]= le.fit_transform(df[col])

for col in categorical_cols:
    print(le.classes_)
```

```python
df['employment']= le.fit_transform(df['employment'])
df['credit history']= le.fit_transform(df['credit history'])
df['creditworthiness']= le.fit_transform(df['creditworthiness'])
df['previous default']= le.fit_transform(df['previous default'])
```

```python
df.head(10)
```

```python
# dropping application id 

df = df.drop("Application id", axis =1)
```

# selecting Target & Features 

```python

X = df.drop("creditworthiness", axis=1)
Y = df['creditworthiness']
```

```python
print(X)
print('Y is \n', Y)
```

```python
# train test split 

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size = 0.20,
    random_state=42
)
```

```python
# check the size 

print(f'Trainning data: {X_train.shape}')
print(f'Testing data: {X_test.shape}')
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
# make prediction 

y_pred = model.predict(X_test)
```

```python
# view prediction 

print(y_pred[:20])
```

### confusion matrix 

```python
# confusion matrix 
cm = confusion_matrix(Y_test, y_pred)

print(cm)
```

#### Accuracy Score 

```python
Acc= accuracy_score(Y_test, y_pred)
print(Acc)
```

### Recall 

```python
recall= recall_score(Y_test, y_pred)
print(recall)
```

```python
precission = precision_score(Y_test, y_pred)
print(precission)
```
