```python
x= df[['Area_sqft', 'Bedrooms','Bathrooms', 'Age_years', 'Distance_to_city_km', 'Parking_spaces']]

y= df['Price_INR']
print(x.shape)
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
# 80 % trainning data and 20% testing data 
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
from sklearn.preprocessing import PolynomialFeatures

# create po;lynomial featues of degree 2

poly = PolynomialFeatures(degree=2)

x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.fit_transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train_poly, y_train)

print("Model trainning completed")
```

```python
y_pred = model.predict(x_test_poly)

# compare actual and predicted price 

result = pd.DataFrame({
    'Actual Price': y_test.values,
    'Predicted Price': y_pred.round(0)
})
result.head(10)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(result['Actual Price'], y_pred)

print(f'R2 is {r2}')

# Display with 3 decimal places
print(f"R2 Score: {r2:.3f}")


# Display as a percentage with 1 decimal place
print(f"Model Accuracy: {r2 * 100:.1f}%")
```

```python
new_house =pd.DataFrame({
    'Area_sqft':[3000],
    'Bedrooms':[4],
    'Bathrooms':[1],
    'Age_years':[25],
    'Distance_to_city_km':[20],
    'Parking_spaces':[1]
})

# convert the new house data into polynomial features
new_house_poly = poly.transform(new_house)

# predict the price
predicted_price = model.predict(new_house_poly)

print(f'Predicted price is : Rs. {predicted_price}')
```

# statistics


```python
# select numerical columns 
import numpy as np
num_col = df.select_dtypes(include=np.number).columns

# check skewness of each numerical column
print("skewness")
print(df[num_col].skew().round(2))

# fill missing values based on skewness

for col in num_col:

    # calculate skewness using available values 
    skew_values = df[col].skew()

    # if skewness is between -0.5 and +0.5 use mean
    if 0.5 <= skew_values <=0.5:
        df[col]=df[col].fillna(df[col].mean())
        print(col, "-> Filled with Mean")

        # otherwize use mediuan

    else:
        df[col] =df[col].fillna(df[col].median())
        print(col, "-> Filled with Median")
```
